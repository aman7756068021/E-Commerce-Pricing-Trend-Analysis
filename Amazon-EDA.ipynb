### Impact of Discounts & Promotions (Box Plot)


```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='category')
plt.xticks(rotation=90)
plt.xlabel('Product Category')
plt.ylabel('Discounted Price')
plt.title('Impact of Discounts on Different Categories')
plt.show()
```

# Market Positioning & Price Competition (Violin Plot)
## Analyzes discounted price distribution across product categories.

```python
plt.figure(figsize=(8,5))
sns.violinplot(x='category', y='discounted_price', data=df)
plt.xticks(rotation=90)
plt.xlabel('Product Category')
plt.ylabel('Discounted Price')
plt.title('Market Positioning & Price Distribution')
plt.show()
```

# Stock & Product Availability Signals (Bar Chart)
## Finds high-rated products with low review counts (potential stock issues).

```python
top_products = df.sort_values(by=['rating', 'rating_count'], ascending=[False, True]).head(10)

plt.figure(figsize=(8,5))
sns.barplot(y=top_products['product_name'], x=top_products['rating_count'])
plt.xlabel('Number of Ratings')
plt.ylabel('Product Name')
plt.title('Products with High Ratings but Low Reviews (Stock Issues)')
plt.show()
```

# Top 10 Discounted Products (Bar Chart)

```python
top_discounted = df.nlargest(10, 'discount_percentage')

# Bar plot
plt.figure(figsize=(10, 5))
sns.barplot(data=top_discounted, x='discount_percentage', y='product_name', palette='coolwarm')
plt.xlabel("Discount Percentage")
plt.ylabel("Product Name")
plt.title("Top 10 Most Discounted Products")
plt.show()
```

# Category-wise Discount Distribution (Box Plot)


```python
top_categories = df['category'].value_counts().nlargest(10).index
df_top_categories = df[df['category'].isin(top_categories)]

plt.figure(figsize=(12, 6))
sns.boxplot(data=df_top_categories, x='category', y='discounted_price', palette='Spectral')
plt.xticks(rotation=45)
plt.xlabel("Category")
plt.ylabel("Discounted Price")
plt.title("Price Distribution Across Top 10 Categories")
plt.show()
```

# Best-Selling Products (Bar Chart)

```python
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')

# Drop NaN values (if any) after conversion
df = df.dropna(subset=['rating_count'])

# Get top 10 best-selling products based on rating count
top_selling = df.nlargest(10, 'rating_count')

# Plot the top-selling products
plt.figure(figsize=(10, 5))
sns.barplot(data=top_selling, x='rating_count', y='product_name', palette='Blues_r')
plt.xlabel("Number of Ratings")
plt.ylabel("Product Name")
plt.title("Top 10 Best-Selling Products")
plt.show()
```
